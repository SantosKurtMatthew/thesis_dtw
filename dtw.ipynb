```python
import numpy as np
```

```python
def dtw(X, Y):
    n = len(X)
    m = len(Y)
    distance_matrix = np.zeros((n, m))
    for i in range(n):
        for j in range(m):
            distance_matrix[i, j] = abs(X[i] - Y[j])
    print(distance_matrix)
    
    cost_matrix = np.zeros((n+1,m+1))
    cost_matrix[0, 0] = 0
    for i in range(n):
        cost_matrix[i+1, 0] = np.inf
    for j in range(m):
        cost_matrix[0, j+1] = np.inf

    print(cost_matrix)
    
    for i in range(1, n+1):
        for j in range(1, m+1):
            cost_matrix[i,j] = distance_matrix[i-1, j-1] + min([cost_matrix[i-1, j-1], 
                                                                cost_matrix[i-1,j], cost_matrix[i, j-1]])

    i = n
    j = m
    optimal_warping_path = [(n,m)]
    while not(i == 1 and j == 1):
        min_val = min([cost_matrix[i-1, j-1], cost_matrix[i-1,j], cost_matrix[i, j-1]])
        if min_val == cost_matrix[i-1, j-1]:
            optimal_warping_path.append((i-1,j-1))
            i -= 1
            j -= 1
        elif min_val == cost_matrix[i-1, j]:
            optimal_warping_path.append((i-1,j))
            i -= 1
        else:
            optimal_warping_path.append((i, j-1))
            j -= 1

    print(cost_matrix)
    return optimal_warping_path

X = [0, 2, 0, 1, 0, 0]
Y = [0, 0, 0.5, 2, 0, 1, 0]

dtw(X, Y)
```
